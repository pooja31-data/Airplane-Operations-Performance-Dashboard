```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("airline_flight_operations.csv")
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=["flight_id"])
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates()
```

```python
df["airline"].unique()
```

```python
df["airline"]=(df["airline"].str.strip().str.title())
```

```python
df["airline"]=(df["airline"].str.strip().replace({"Airindia" : "Air India",
                                                 "Akasaair" : "Akasa Air",
                                                 "Spicejet" : "Spice Jet"}))
```

```python
df["airline"].unique()
```

```python
df["origin"].unique()
```

```python
df["origin"]=(df["origin"].str.strip().replace({
    "Bangluru" : "Bengaluru"}))
```

```python
df["destination"].unique()
```

```python
df["destination"]=(df["destination"].str.strip().str.title())
```

```python
df["destination"]=(df["destination"].str.strip().replace({
    "Bangalore" : "Bengaluru"}))
```

```python
df.isnull().sum()
```

```python
df["cancelled"].unique()
```

```python
df[df["cancellation_reason"].isna()][["cancelled","cancellation_reason"]].head()
```

```python
df.loc[(df["cancelled"]==False) & (df["cancellation_reason"].isna()),"cancellation_reason"]="Not Cancelled"
```

```python
df.loc[(df["cancelled"]==True) & (df["cancellation_reason"].isna()),"cancellation_reason"]="Unknown"
```

```python
df["cancellation_reason"].unique()
```

```python
df.isnull().sum()
```

```python
df["aircraft_type"]=df["aircraft_type"].str.strip()
```

```python
df["aircraft_type"]=df["aircraft_type"].fillna("Unknown")
```

```python
df["aircraft_type"].unique()
```

```python
df.isnull().sum()
```

```python
df["passengers"]=pd.to_numeric(df["passengers"],errors="coerce")
```

```python
df["passengers"]=df["passengers"].fillna(df["passengers"].median())
df["passengers"]=df["passengers"].astype(int)
```

```python
df.isnull().sum()
```

```python
df["weather_condition"].unique()
```

```python
df["weather_condition"]=(df["weather_condition"].str.strip().str.title())
```

```python
df["weather_condition"]=df["weather_condition"].fillna("Unknown")
```

```python
df["ticket_price"]=pd.to_numeric(df["ticket_price"],errors="coerce")
```

```python
df["ticket_price"]=df["ticket_price"].fillna(df["ticket_price"].median())
df["ticket_price"]=df["ticket_price"].astype(int)
```

```python
df["ticket_price"].unique()
```

```python
df["airline"]=df["airline"].fillna("Unknown")
df["origin"]=df["origin"].fillna("Unknown")
df["destination"]=df["destination"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["cancellation_reason"].unique()
```

```python
df["cancelled"].unique()
```

```python
df["cancelled"]=(df["cancelled"].str.strip().str.title().replace({
    "1" : True,
    "True" : True,
    "Yes" : True,
    "0" : False,
    "No" : False,
    "False" : False}))
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
df["cancellation_reason"].unique()
```

```python
df.loc[:,"cancellation_reason"]=df["cancellation_reason"].fillna("Unknown")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.loc[:,"status"]=(df["status"].str.strip().str.title())
```

```python
df["status"].unique()
```

```python
df["airport_congestion"].unique()
```

```python
df["weather_condition"].unique()
```

```python
df.info()
```

```python
df["flight_date"].unique()
```

```python
test_dates = (
    df["flight_date"]
    .astype("string")
    .str.strip()
    .str.replace("/", "-", regex=False)
)

test_dates = pd.to_datetime(
    test_dates,
    format="mixed",
    errors="raise"
)
```

```python
df["flight_date"]=test_dates
```

```python
df.info()
```

```python
df["scheduled_departure"].unique()
```

```python
test_scheduled_departure = pd.to_datetime(
    df["scheduled_departure"]
        .astype("string")
        .str.strip()
        .str.replace("/", "-", regex=False),
    format="mixed",
    errors="raise"
)
```

```python
df["scheduled_departure"]=test_scheduled_departure
```

```python
df.info()
```

```python
test_actual_departure = pd.to_datetime(
    df["actual_departure"]
        .astype("string")
        .str.strip()
        .str.replace("/", "-", regex=False),
    format="mixed",
    errors="raise"
)
```

```python
df["actual_departure"]=test_actual_departure
```

```python
df.info()
```

```python
test_scheduled_arrival = pd.to_datetime(
    df["scheduled_arrival"]
        .astype("string")
        .str.strip()
        .str.replace("/", "-", regex=False),
    format="mixed",
    errors="raise"
)
```

```python
df["scheduled_arrival"]=test_scheduled_arrival
```

```python
test_actual_arrival = pd.to_datetime(
    df["actual_arrival"]
        .astype("string")
        .str.strip()
        .str.replace("/", "-", regex=False),
    format="mixed",
    errors="raise"
)
```

```python
df["actual_arrival"]=test_actual_arrival
```

```python
df.info()
```

```python
df["arrival_delay"].unique()
```

```python
df["departure_delay"]=pd.to_numeric(
    df["departure_delay"],errors="coerce")
df["departure_delay"]=df["departure_delay"].fillna(df["departure_delay"].median())                                    
```

```python
df["arrival_delay"]=pd.to_numeric(
    df["arrival_delay"],errors="coerce")
df["arrival_delay"]=df["arrival_delay"].fillna(df["arrival_delay"].median())                                    
```
